```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. 데이터 세팅
# -----------------------------
text_types = ["Statement", "Minutes", "S+M"]
x = np.arange(len(text_types))   # [0, 1, 2]
width = 0.35                     # 막대 폭

# Alpha (최고 성능 기준: Statement / Minutes / S+M)
alpha_acc = [0.647, 0.688, 0.647]
alpha_f1  = [0.588, 0.560, 0.626]

# Beta (최고 성능 기준: Statement / Minutes / S+M)
beta_acc = [0.636, 0.647, 0.722]
beta_f1  = [0.599, 0.650, 0.723]

ymin, ymax = 0.0, 0.8  # 공통 y축 범위

# -----------------------------
# 2. Figure & Subplots 생성
# -----------------------------
fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True)
ax_alpha, ax_beta = axes

# -----------------------------
# 2-1. Alpha 그래프
# -----------------------------
ax_alpha.bar(x - width/2, alpha_acc, width, label="Accuracy", alpha=0.85)
ax_alpha.bar(x + width/2, alpha_f1,  width, label="F1-score", alpha=0.85)

ax_alpha.set_xticks(x)
ax_alpha.set_xticklabels(text_types)
ax_alpha.set_ylim(ymin, ymax)
ax_alpha.set_title("Alpha (ΔPC1) 성능 비교")
ax_alpha.set_ylabel("Score")
ax_alpha.legend(loc="upper left")

# 값 라벨링 (선택)
for i in range(len(x)):
    ax_alpha.text(x[i] - width/2, alpha_acc[i] + 0.01, f"{alpha_acc[i]:.2f}",
                  ha="center", va="bottom", fontsize=9)
    ax_alpha.text(x[i] + width/2, alpha_f1[i]  + 0.01, f"{alpha_f1[i]:.2f}",
                  ha="center", va="bottom", fontsize=9)

# -----------------------------
# 2-2. Beta 그래프
# -----------------------------
ax_beta.bar(x - width/2, beta_acc, width, label="Accuracy", alpha=0.85)
ax_beta.bar(x + width/2, beta_f1,  width, label="F1-score", alpha=0.85)

ax_beta.set_xticks(x)
ax_beta.set_xticklabels(text_types)
ax_beta.set_ylim(ymin, ymax)
ax_beta.set_title("Beta (ΔPC2) 성능 비교")
ax_beta.legend(loc="upper left")

# 값 라벨링 (선택)
for i in range(len(x)):
    ax_beta.text(x[i] - width/2, beta_acc[i] + 0.01, f"{beta_acc[i]:.2f}",
                 ha="center", va="bottom", fontsize=9)
    ax_beta.text(x[i] + width/2, beta_f1[i]  + 0.01, f"{beta_f1[i]:.2f}",
                 ha="center", va="bottom", fontsize=9)

plt.tight_layout()
plt.show()
```

```python
''' 
*Alpha(ΔPC1)**에서는 Minutes 단독 사용이 정확도는 가장 높지만, 성명문과 의사록을 함께 쓴 S+M이 F1-score가 가장 높아 두 텍스트를 합쳤을 때 매파·비둘기 균형 있게 예측한다는 걸 보여준다.

**Beta(ΔPC2)**에서는 S+M이 정확도와 F1-score 모두 0.72 수준으로 Statement·Minutes 단독보다 뚜렷이 높아, 기울기 서프라이즈는 두 문서의 정보를 함께 쓸 때 가장 잘 설명되는 것으로 나타난다.

전반적으로 Statement만 쓴 경우가 가장 약하고, Minutes와 S+M으로 갈수록 성능이 개선되어 회의 전문(의사록)과 성명문을 동시에 보는 것이 정책 서프라이즈 정보를 더 풍부하게 반영한다는 해석이 가능하다.


S+M이 가장 높은 이유
성명문은 시장이 바로 보는 “헤드라인 톤”, 의사록은 뒤늦게 공개되는 “회의 내용·논리·우려사항”이 담겨 있어서, 둘을 같이 쓰면 즉각적 신호 + 상세한 맥락이 동시에 입력에 들어간다. 그래서 Alpha/Beta 모두에서 S+M이 가장 높은 성능을 보이는 건, 두 문서가 서로 다른 정보를 보완적으로 담고 있고, 서프라이즈 라벨과의 정합성이 가장 풍부해진 결과로 볼 수 있다.

Minutes가 Statement보다 일관되게 더 좋은 이유
Minutes는 분량이 훨씬 길고, 개별 위원의 우려·논쟁·리스크 인식 등이 더 구체적으로 드러나기 때문에, 모델이 **매파/비둘기 방향을 구분할 수 있는 힌트(단어·표현·토픽)**를 더 많이 학습할 수 있다. 반면 Statement는 짧고 정제된 문장 위주라 정보량이 상대적으로 적어, “회의의 전체 분위기·논리 구조”보다는 표면적 톤만 담겨 있는 문서라서 성능이 낮게 나온 것으로 해석할 수 있다

'''
```
